```python
# PASO 1: Instalar graph-tool con conda
# ⚠️ IMPORTANTE: Esta celda instalará condacolab y REINICIARÁ el runtime
# Después del reinicio, ejecuta las celdas siguientes (3, 4, 5, 6)

import os
print("Instalando condacolab y graph-tool...")
os.system("pip install -q condacolab")
import condacolab
condacolab.install()
os.system("conda install -c conda-forge -y graph-tool")
print("✓ graph-tool instalado. El runtime se reiniciará ahora.")
```

```python
# PASO 2: Verificar GPU
import torch
print(f"GPU disponible: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"Memoria: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
```

```python
# PASO 2: Clonar repositorio
import subprocess
import os

repo_dir = "/content/CoPHo"
subprocess.run(["git", "clone", "https://github.com/abarrenap/CoPHo.git", repo_dir])
os.chdir(repo_dir)
print(f"✓ Repositorio clonado en: {repo_dir}")
```

```python
# PASO 2: Ejecutar entrenamiento
os.chdir(f"{repo_dir}/src")

cmd = f"python main.py --config-name=config_mis_persisthomo general.name=mis_dimacs_300"
print(f"Ejecutando: {cmd}\n")
result = subprocess.run(cmd, shell=True, capture_output=True, text=True)

# Mostrar output
print("STDOUT:")
print(result.stdout)
print("\nSTDERR:")
print(result.stderr)
print(f"\nCódigo de salida: {result.returncode}")
```
